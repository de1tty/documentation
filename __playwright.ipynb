# Как работает код?

---

**1. Навигация**

Обычно тест начинается с перехода на страницу по указанной ссылке, так как только после этого тест может взаимодействовать с элементами страницы

```
page.goto("https://playwright.dev/")
```
Playwright дождется, когда страница загрузится

**2. Взаимодействие**

Перед началом действий, нужно найти элементы. Для этого Playwright использует *Locators API*. *Локаторы* — это способ найти элемент(ы) на странице в любой момент. Playwright будет ждать, пока элемент станет доступным для действий


```
# Создает локатор
get_started = page.get_by_role("link", name="Get started")

# Нажимает на него
get_started.click()
```
Обычно это пишут в одну строчку:

```
page.get_by_role("link", name="Get started").click()
```


# Основные действия и их значения


---


```
Действие                                   Значение
locator.check()      -------------->       Установите флажок ввода.
locator.click()      -------------->       Нажмите на элемент.
locator.uncheck()    -------------->       Снимите флажок с поля ввода.
locator.hover()      -------------->       Наведите курсор мыши на элемент.
locator.fill()       -------------->       Заполните поле формы, введите текст.
locator.focus()      -------------->       Сфокусируйте элемент.
locator.press()      -------------->       Нажмите одну клавишу.
locator.set_input_files()  ------------>   Выберите файлы для загрузки.
locator.select_option()   ------------->   Выберите вариант в выпадающем списке.
```







**Утверждения**

Они ждут выполнения ожидаемого условия, их использование позволяет сделать тесты нестабильными и устойчивыми. Например, этот код будет ждать, пока страница не получит заголовок, содержащий слово «Playwright»:

```
import re
from playwright.sync_api import expect

expect(page).to_have_title(re.compile("Playwright"))
```



# Популярные утверждения и их значения


---



```
Утверждение                                                Значение
expect(locator).to_be_checked()       -------------->      Флажок установлен
expect(locator).to_be_enabled()       -------------->      Управление включено
expect(locator).to_be_visible()       -------------->      Элемент виден
expect(locator).to_contain_text()     -------------->      Элемент содержит текст
expect(locator).to_have_attribute()   -------------->      Элемент имеет атрибут
expect(locator).to_have_count()       -------------->      Список элементов имеет заданную длину
expect(locator).to_have_text()        -------------->      Элемент соответствует тексту
expect(locator).to_have_value()       -------------->      Входной элемент имеет значение
expect(page).to_have_title()          -------------->      Страница имеет заголовок
expect(page).to_have_url()            -------------->      Страница имеет URL
```



# Запуск Codegen. Что это и как работает?
---
эту команду используют для запуска генератора тестов, вот как она выглядит:

```
playwright codegen demo.playwright.dev/todomvc
```
Важно вставить URL-адрес сайта, для которого хотите провести тестирование

**Как записывается тест?**

Запустите codegen и выполните любые действия в браузере. Playwright автоматически генерирует код для ваших взаимодействий. Codegen анализирует отображаемую страницу и рекомендует наиболее подходящий локатор, отдавая приоритет локаторам по роли, тексту и идентификатору теста. Если несколько элементов соответствуют локатору, генератор улучшает его, чтобы конкретно идентифицировать целевой элемент, чтобы они не путались между собой, это помогает тестам работать лучше и точнее

**Что можно записать с помощью генератора тестов?**


*   Любые взаимодействия (например клик, заполнение страницы)
*   Утвеждения взаимодействия: `'assert visibility'`чтобы убедиться, что элемент виден, `'assert text'`чтобы убедиться, что элемент содержит определенный текст, `'assert value'`чтобы убедиться, что элемент имеет определенное значение

Когда вы закончите тест, нажмите `'record'`, чтобы остановить запись, а затем `'copy'`, чтобы скопировать сгенерированный код в редактор, так же используйте кнопку `'clear'`, чтобы очистить код и начать запись заново, когда закончите, закройте окно Playwright Inspector или отмените команду в терминале

# Создание локаторов
---
Так же можно создавать локаторы с помощью генератора тестов:


*  Нажмите `'Record'`, чтобы остановить запись, и появится `'Pick Locator'`
*   Нажмите `'Pick Locator'` кнопку и наведите мышкой на элементы в окне браузера, чтобы увидеть локатор, выделенный под каждым элементом


*   Щелкните элемент, который вы хотите найти, и код для этого локатора появится на игровой площадке `locator` рядом с кнопкой Выбора локатора
*   Отредактируйте локатор на игровой площадке `locator`, чтобы настроить его, и вы увидите соответствующий элемент, выделенный в окне браузера

*   Используйте кнопку копировать, чтобы скопировать локатор и вставить его в свой код
